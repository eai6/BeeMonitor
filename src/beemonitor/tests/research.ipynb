```python
# save the dictionary objects of all the results as JSON (handles Timestamps/numpy types)
import json
import numpy as np
import datetime
import pandas as pd
import os

def to_serializable(o):
    if isinstance(o, pd.DataFrame):
        return o.to_dict(orient='records')
    if isinstance(o, dict):
        return {str(k): to_serializable(v) for k, v in o.items()}
    if isinstance(o, (list, tuple, set)):
        return [to_serializable(x) for x in o]
    if isinstance(o, np.ndarray):
        return to_serializable(o.tolist())
    if isinstance(o, (np.integer,)):
        return int(o)
    if isinstance(o, (np.floating,)):
        return float(o)
    if isinstance(o, (pd.Timestamp, datetime.datetime)):
        return o.isoformat()
    if hasattr(o, "to_dict") and not isinstance(o, (str, bytes)):
        try:
            return to_serializable(o.to_dict())
        except Exception:
            return str(o)
    try:
        json.dumps(o)
        return o
    except Exception:
        return str(o)

def _json_default(o):
    # fallback for json.dump in case something slipped through
    if isinstance(o, (pd.Timestamp, datetime.datetime)):
        return o.isoformat()
    if isinstance(o, np.integer):
        return int(o)
    if isinstance(o, np.floating):
        return float(o)
    if isinstance(o, np.ndarray):
        return o.tolist()
    if hasattr(o, "to_dict"):
        try:
            return o.to_dict()
        except Exception:
            return str(o)
    return str(o)

results_serializable = {}
for video_path, analysis_results in results1.items():
    if analysis_results is None:
        continue
    key = os.path.basename(video_path)
    obj = analysis_results.to_dict() if hasattr(analysis_results, "to_dict") else analysis_results
    results_serializable[key] = to_serializable(obj)

# with open("cvpr_evaluation_results_v3.json", "w") as f:
#     json.dump(results_serializable, f, indent=2, default=_json_default)


with open("cvpr_evaluation_results_v7_medium.json", "w") as f:
    json.dump(results_serializable, f, indent=2, default=_json_default)
```

```python
# get all events from results1
import pandas as pd
import os
def getAllEvents(results):
    df_events = pd.DataFrame()
    for video_file, analysis_results in results.items():
        print(video_file)
        print(os.path.basename(video_file))
        if analysis_results is not None:
            df_events_temp = analysis_results.events.copy()
            df_events_temp['filename'] = video_file
            df_events_temp['video'] = os.path.basename(video_file)
            df_events = pd.concat([df_events, df_events_temp], ignore_index=True)
    return df_events
```

```python

from typing import List
import numpy as np

def _create_composite_frame(self, frames: List[np.ndarray]) -> np.ndarray:
    """Create max composite from frame buffer for motion detection."""
    if not frames:
        return None
    return np.maximum.reduce(frames)
```

```python
buffer = []
while True:
    ret, frame = cap.read()
    if not ret:
        break
    buffer.append(frame)
    if len(buffer) == 10:
        composite_frame = np.maximum.reduce(buffer)
        cv.imshow("Composite Frame", composite_frame)
        cv.waitKey(0)
        buffer = []
cap.release()
cv.destroyAllWindows()
```
